# Neovide Notebook 演示
先按 空格 j i 启动内核，再在代码单元格按 Shift+Enter 执行。使用 ]j / [j 跳转，i 编辑当前格，Esc 返回浏览，:w 保存。

```python
scores = [82, 91, 76, 95, 88]
print('Notebook 已在 Neovide 中运行！')
print(f'成绩：{scores}')
```

```python
average = sum(scores) / len(scores)
print(f'平均分：{average:.1f}')
print(f'最高分：{max(scores)}')
```
